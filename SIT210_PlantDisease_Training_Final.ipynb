### 2.2 Prepare Dataset (Train / Val / Test Split)

Split dataset into 70% train / 15% val / 15% test using stratified splitting.

```python
import shutil
from sklearn.model_selection import train_test_split

PREPARED_DIR = '/content/prepared'

for split in ['train', 'val', 'test']:
    for cls in all_classes:
        os.makedirs(f'{PREPARED_DIR}/{split}/{cls}', exist_ok=True)

for cls in all_classes:
    src = f'{dataset_base}/{cls}'
    images = os.listdir(src)
    train_imgs, temp = train_test_split(images, test_size=0.3, random_state=42)
    val_imgs, test_imgs = train_test_split(temp, test_size=0.5, random_state=42)

    for img in train_imgs:
        shutil.copy(f'{src}/{img}', f'{PREPARED_DIR}/train/{cls}/{img}')
    for img in val_imgs:
        shutil.copy(f'{src}/{img}', f'{PREPARED_DIR}/val/{cls}/{img}')
    for img in test_imgs:
        shutil.copy(f'{src}/{img}', f'{PREPARED_DIR}/test/{cls}/{img}')

print(f"✅ Dataset prepared: {len(all_classes)} classes split into train/val/test")
```

## 3. Build Model

Using **MobileNetV2** pre-trained on ImageNet as the backbone — ideal for edge deployment due to its small size and efficient depthwise separable convolutions.

**Architecture:**
- MobileNetV2 backbone (frozen) → GlobalAveragePooling → Dropout(0.3) → Dense(256) → Dropout(0.2) → Softmax(38)

**Why MobileNetV2?**
- Only ~4MB when quantized → fits comfortably on Raspberry Pi
- Fast inference (~650ms on Pi 3B+ CPU)
- Strong ImageNet features transfer well to plant leaf classification

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau

IMG_SIZE = 224
BATCH_SIZE = 32
NUM_CLASSES = 38

print(f"✅ TensorFlow {tf.__version__}")
print(f"✅ GPU: {tf.config.list_physical_devices('GPU')}")

base_model = MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False  # Freeze for Phase A

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dropout(0.3)(x)
x = Dense(256, activation='relu')(x)
x = Dropout(0.2)(x)
output = Dense(NUM_CLASSES, activation='softmax')(x)

model = Model(inputs=base_model.input, outputs=output)
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
print(f"✅ Model built — {model.count_params():,} total parameters")
```

## 4. Data Augmentation & Generators

Training data is augmented with random flips, rotations, and zooms to improve generalization and reduce overfitting.

```python
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    horizontal_flip=True,
    zoom_range=0.2,
    fill_mode='nearest'
)
val_datagen = ImageDataGenerator(rescale=1./255)  # No augmentation for val/test

train_gen = train_datagen.flow_from_directory(
    f'{PREPARED_DIR}/train',
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)
val_gen = val_datagen.flow_from_directory(
    f'{PREPARED_DIR}/val',
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical'
)
test_gen = val_datagen.flow_from_directory(
    f'{PREPARED_DIR}/test',
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False  # Keep order for evaluation
)
print(f"✅ Train: {train_gen.n} | Val: {val_gen.n} | Test: {test_gen.n} images")
```

## 5. Training

### Phase A — Transfer Learning (Frozen Base)

Train only the classification head while MobileNetV2 weights are frozen.  
This is fast (~5–10 min on T4 GPU) and establishes a strong baseline.

```python
callbacks = [
    ModelCheckpoint(
        f'{PROJECT_DIR}/models/best_model.h5',
        monitor='val_accuracy', save_best_only=True, verbose=1
    ),
    EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=1e-7, verbose=1)
]

print("🚀 Training Phase A (frozen base)...")
history_a = model.fit(
    train_gen,
    epochs=10,
    validation_data=val_gen,
    callbacks=callbacks,
    verbose=1
)
print(f"✅ Phase A complete — best val_accuracy: {max(history_a.history['val_accuracy']):.4f}")
```

### Phase B — Fine-Tuning

Unfreeze the last 34 layers of MobileNetV2 and train with a lower learning rate (1e-4).  
This lets the backbone adapt to plant disease features.

```python
# Unfreeze last 34 layers of the model
for layer in model.layers:
    layer.trainable = False
for layer in model.layers[-34:]:
    layer.trainable = True

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

trainable_count = sum(1 for l in model.layers if l.trainable)
print(f"✅ {trainable_count} layers trainable for fine-tuning")
print("🚀 Training Phase B (fine-tuning)...")

history_b = model.fit(
    train_gen,
    epochs=10,
    validation_data=val_gen,
    callbacks=callbacks,
    verbose=1
)
print(f"✅ Phase B complete — best val_accuracy: {max(history_b.history['val_accuracy']):.4f}")
```

## 6. Export to TensorFlow Lite

Convert the trained Keras model to TFLite with **INT8 dynamic range quantization**.

Benefits for edge deployment:
- Model size reduced from ~14MB → ~4MB
- Faster inference on Pi 3B+ CPU (no GPU required)
- Lower memory footprint

```python
import tensorflow as tf

# Load best saved model
model = tf.keras.models.load_model(f'{PROJECT_DIR}/models/best_model.h5')
print("✅ Best model loaded!")

# Convert with INT8 quantization
converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_model = converter.convert()

# Save TFLite model
tflite_path = f'{PROJECT_DIR}/models/plant_disease_model.tflite'
with open(tflite_path, 'wb') as f:
    f.write(tflite_model)

size_mb = len(tflite_model) / (1024 * 1024)
print(f"✅ TFLite model saved! Size: {size_mb:.1f} MB")
print(f"   Path: {tflite_path}")
```

## 7. Inference Demo

Run the TFLite model on 9 random test images to verify predictions.  
This replicates the exact inference pipeline used on the Raspberry Pi.

```python
import numpy as np
import matplotlib.pyplot as plt
import json, random, os
from tensorflow.keras.preprocessing import image

# Load class names and TFLite model
with open(f'{PROJECT_DIR}/models/class_names.json', 'r') as f:
    class_names = json.load(f)

interpreter = tf.lite.Interpreter(
    model_path=f'{PROJECT_DIR}/models/plant_disease_model.tflite'
)
interpreter.allocate_tensors()
input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

def predict_image(img_path):
    img = image.load_img(img_path, target_size=(224, 224))
    img_array = image.img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0).astype(np.float32)

    interpreter.set_tensor(input_details[0]['index'], img_array)
    interpreter.invoke()
    output = interpreter.get_tensor(output_details[0]['index'])[0]

    predicted_idx = str(np.argmax(output))
    confidence = output[int(predicted_idx)] * 100
    predicted_class = class_names[predicted_idx]
    return predicted_class, confidence

# Visualise 9 random test images
test_dir = f'{PROJECT_DIR}/dataset/prepared/test'
fig, axes = plt.subplots(3, 3, figsize=(12, 12))
axes = axes.flatten()

for i in range(9):
    cls = random.choice(os.listdir(test_dir))
    img_file = random.choice(os.listdir(f'{test_dir}/{cls}'))
    img_path = f'{test_dir}/{cls}/{img_file}'

    predicted, confidence = predict_image(img_path)
    correct = predicted == cls
    color = 'green' if correct else 'red'

    img_display = image.load_img(img_path, target_size=(224, 224))
    axes[i].imshow(img_display)
    axes[i].set_title(
        f'Actual: {cls[:25]}\nPred: {predicted[:25]}\nConf: {confidence:.1f}%',
        color=color, fontsize=8
    )
    axes[i].axis('off')

plt.suptitle('TFLite Inference Demo — Green = Correct, Red = Wrong', fontsize=13, y=1.01)
plt.tight_layout()
plt.savefig(f'{PROJECT_DIR}/models/inference_demo.png', bbox_inches='tight')
plt.show()
print("✅ Inference demo complete!")
```
